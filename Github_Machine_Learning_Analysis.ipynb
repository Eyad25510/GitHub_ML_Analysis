```python
import requests
import pandas as pd
```

```python
url = "https://api.github.com/search/repositories?q=machine+learning&sort=stars&order=desc&per_page=100"
```

```python
response = requests.get(url)
response.status_code
```

```python
data = response.json()
type(data)
```

```python
data.keys()
```

```python
len(data["items"])
```

```python
df = pd.DataFrame(data["items"])
```

```python
df.head()
```

```python
df.shape
```

```python
df.columns
```

```python
df.info()
```

```python
df.tail()
```

```python
required_columns = [
    "name",
    "owner",
    "language",
    "stargazers_count",
    "forks_count",
    "watchers_count",
    "open_issues_count",
    "created_at",
    "updated_at",
    "license"
]

df = df[required_columns]
```

```python
df.head()
```

```python
df.shape
```

```python
df["owner"] = df["owner"].apply(lambda x: x["login"] if x else None)
```

```python
df["owner"].head()
```

```python
df["license"] = df["license"].apply(lambda x: x["name"] if x else None)
```

```python
df["license"].head()
```

```python
df.isnull().sum()
```

```python
df["language"] = df["language"].fillna("Unknown")
df["license"] = df["license"].fillna("No License")
```

```python
df.isnull().sum()
```

```python
df.duplicated().sum()
```

```python
df = df.drop_duplicates()
```

```python
df["created_at"] = pd.to_datetime(df["created_at"])
df["updated_at"] = pd.to_datetime(df["updated_at"])
```

```python
df.dtypes
```

```python
df = df.rename(columns={
    "stargazers_count": "stars",
    "forks_count": "forks",
    "watchers_count": "watchers",
    "open_issues_count": "open_issues",
    "created_at": "created_date",
    "updated_at": "updated_date"
})
```
